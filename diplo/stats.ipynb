```python
import json
import numpy as np
```

```python
metrics_to_evaluate = [
    "adjective_ratio", 
    "function_words", 
    'content_words',
    "sentences_per_paragraph",
    "words_per_sentence", 
    "words", 
    "sentences", 
    "adverbs",
    "logic_operators",
    "concretude_mean",
    "brunet",
    "honore",
    "pronoun_ratio", 
    "simple_word_ratio", 
    "long_sentence_ratio",

    # Ambiguidade
    "adjectives_ambiguity",
    "adverbs_ambiguity",
    "nouns_ambiguity",
    "verbs_ambiguity",
    "content_words_ambiguity",

    # Diversidade (agrupadas)
    "adjective_diversity_ratio", 
    "adverbs_diversity_ratio", 
    "content_word_diversity",
    "function_word_diversity",
    "verbal_time_moods_diversity", 
    "indefinite_pronouns_diversity", 
    "preposition_diversity",
    "pronoun_diversity", 
    "punctuation_diversity", 
    "verb_diversity"
]
```

```python
def load_metrics(filename):
    with open(filename, "r", encoding="utf-8") as f:
        data = json.load(f)
    values = {metric: [] for metric in metrics_to_evaluate}
    for entry in data.values():
        metrics = entry.get("metrics", {})
        for metric in metrics_to_evaluate:
            if metric in metrics:
                values[metric].append(metrics[metric])
    return {
        m: (np.mean(v), np.std(v)) if v else (None, None)
        for m, v in values.items()
    }

enem_stats = load_metrics("enem_metrics.json")
candidates_stats = load_metrics("candidates_metrics.json")

print(f"{'Metric':<35} {'ENEM avg ± std':>15} {'':^3} {'Diplomatas avg ± std':>25}")
print("-" * 95)

for metric in metrics_to_evaluate:
    enem_mean, enem_std = enem_stats[metric]
    cand_mean, cand_std = candidates_stats[metric]

    if enem_mean is None or cand_mean is None:
        print(f"{metric:<35} {'N/A':>30} {'?':^3} {'N/A':>25}")
        continue

    symbol = ">" if enem_mean > cand_mean else "<" if enem_mean < cand_mean else "="
    print(f"{metric:<35} "
          f"{enem_mean:8.5f} ± {enem_std:7.5f} {symbol:^3} "
          f"{cand_mean:8.5f} ± {cand_std:7.5f}")
```
